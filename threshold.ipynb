```python
import numpy as np
from collections import defaultdict
from sklearn.preprocessing import normalize
from sklearn.metrics.pairwise import cosine_similarity

def compute(threshold, X_gallery, y_gallery,
            X_known, y_known, X_unknown, y_unknown, seed):

    # 1. Group gallery embeddings by subject
    subject_features = defaultdict(list)

    for feat, label in zip(X_gallery, y_gallery):
        subject_features[label].append(feat)

    # 2. Calculate subject prototypes
    prototype_labels = np.array(list(subject_features.keys()))

    prototype_vectors = np.stack([
        np.mean(subject_features[label], axis=0)
        for label in prototype_labels
    ])

    # 3. Normalize prototypes
    prototype_vectors = normalize(prototype_vectors, norm='l2')

    # 4. Identification function
    def identify(X):
        sims = cosine_similarity(X, prototype_vectors)
        best_idx = np.argmax(sims, axis=1)
        best_scores = sims[np.arange(len(X)), best_idx]
        predictions = prototype_labels[best_idx].copy()
        predictions[best_scores < threshold] = -1
        return predictions, best_scores

    # 5. Known probe evaluation
    known_predictions, known_scores = identify(X_known)

    DIR = 100 * np.mean(known_predictions == y_known)

    # 6. Unknown probe evaluation
    unknown_predictions, unknown_scores = identify(X_unknown)

    URR = 100 * np.mean(unknown_predictions == -1)

    print(f"Seed: {seed}")
    # print(f"Prototypes: {len(prototype_labels)}")
    print(f"DIR: {DIR:.4f}%")
    print(f"URR: {URR:.4f}%")
```
